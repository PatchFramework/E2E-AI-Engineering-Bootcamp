# Tracing with Human Feedback

```python
# deps
from langsmith import Client
```

### Attach Feedback to trace in langsmith

```python
client = Client()
```

```python
RUN_ID = "befe6fdd-82b1-45c6-a117-f273d9e439fa"
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="feedback",
    score=0,
    feedback_source_type="api"
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment",
    value="This is a long form feedback test",
    feedback_source_type="api"
)
```
